```python
celsius = [0, 10, 20, 30, 40]

fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))

print(fahrenheit)
```

```python
lista1 = [10, 20, 30, 40]
lista2 = [5, 10, 15, 20]

suma = list(map(lambda a, b: a + b, lista1, lista2))

print(suma)
```

```python
radios = [2, 3, 4, 5]

areas = list(map(lambda r: 3.1416 * r**2, radios))

print(areas)
```

```python
productos = {
    "pan": 1000,
    "leche": 2500,
    "café": 5000
}

resultado = list(map(lambda p: (p[0], p[1] * 1.10), productos.items()))

print(resultado)
```

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

resultado = list(map(lambda fila: list(map(lambda n: n * 10, fila)), matriz))

print(resultado)
```
