```python
palabras = ["pera", "manzana", "uva", "banano", "fresa"]

resultado = sorted(palabras)

print(resultado)
```

```python
tuplas = [(3, 5), (1, 8), (2, 4), (4, 6)]

resultado = sorted(tuplas, key=lambda x: x[0])

print(resultado)
```

```python
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]

resultado = sorted(productos, key=lambda x: (-x[1], x[0]))

print(resultado)
```

```python
palabras = ["computador", "sol", "casa", "universidad", "perro"]

resultado = sorted(palabras, key=lambda x: len(x))

print(resultado)
```

```python
numeros = [7, 4, 9, 2, 6, 3, 8, 5]

resultado = sorted(numeros, key=lambda x: (x % 2, x))

print(resultado)
```

```python
palabra = "programacion"

frecuencia = {}

for letra in palabra:
    if letra in frecuencia:
        frecuencia[letra] = frecuencia[letra] + 1
    else:
        frecuencia[letra] = 1

resultado = sorted(frecuencia.items(), key=lambda x: x[1], reverse=True)

print(resultado)
```

```python
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]

resultado = sorted(empleados, key=lambda x: (x[1], x[2]))

print(resultado)
```
