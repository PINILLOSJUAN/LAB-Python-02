```python
numeros = [5, 10, 15, 20, 25, 30]

a = [(n, n**2) for n in numeros]
print("A:", a)

b = [n**3 for n in numeros if n > 10]
print("B:", b)

c = [n for n in numeros if n % 2 == 0]
print("C:", c)

d = ["valor: " + str(n) for n in numeros]
print("D:", d)

temperaturas = [0, 10, 20, 30, 40]
e = [(c * 9/5) + 32 for c in temperaturas]
print("E:", e)

palabras = ["manzana", "sol", "computador", "casa", "programacion"]
f = [p for p in palabras if len(p) > 5]
print("F:", f)

g = [p[0] for p in palabras]
print("G:", g)

h = ["par" if n % 2 == 0 else "impar" for n in numeros]
print("H:", h)

matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
i = [n for fila in matriz for n in fila]
print("I:", i)
```
