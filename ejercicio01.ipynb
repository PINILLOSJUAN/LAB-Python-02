```python
a = lambda numero: numero % 3 == 0

b = lambda numero: numero ** 3

c = lambda numero1, numero2: numero1 * numero2

d = lambda numero1, numero2: max(numero1, numero2)

e = lambda palabra: palabra.lower().startswith("a")

f = lambda celsius: (celsius * 9/5) + 32


print("A:", a(9))
print("B:", b(3))
print("C:", c(4, 5))
print("D:", d(8, 12))
print("E:", e("Ana"))
print("F:", f(30))
```

```python
funciones = [
    lambda numero: numero * 2,
    lambda numero: numero + 10,
    lambda numero: numero ** 2
]

numero = int(input("Ingrese un número: "))

for funcion in funciones:
    print(funcion(numero))
```
