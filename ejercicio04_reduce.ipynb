```python
from functools import reduce

numeros = [2, 3, 4, 5]

resultado = reduce(lambda a, b: a * b, numeros)

print("Multiplicación:", resultado)
```

```python
from functools import reduce

palabras = ["Hola", " ", "mundo", " ", "Python"]

resultado = reduce(lambda a, b: a + b, palabras)

print("Palabras concatenadas:", resultado)
```

```python
from functools import reduce

numeros = [10, 25, 8, 40, 15]

def mayor(a, b):
    if a > b:
        return a
    else:
        return b

resultado = reduce(mayor, numeros)

print("El número mayor es:", resultado)
```
