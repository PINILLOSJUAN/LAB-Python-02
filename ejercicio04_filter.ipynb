```python
palabras = ["casa", "computador", "sol", "perro", "universidad"]

resultado = list(filter(lambda p: len(p) > 4, palabras))

print(resultado)
```

```python
lista = [10, None, 20, None, 30, 40, None]

resultado = list(filter(lambda x: x is not None, lista))

print(resultado)
```

```python
palabras = ["avion", "perro", "elefante", "casa", "oso", "uva"]

resultado = list(filter(lambda p: p[0].lower() in "aeiou", palabras))

print(resultado)
```

```python
palabras = ["oso", "casa", "radar", "perro", "reconocer"]

resultado = list(filter(lambda p: p == p[::-1], palabras))

print(resultado)
```

```python
numeros = [15, 20, 35, 40, 55, 60, 75]

resultado = list(filter(lambda n: n % 10 == 5, numeros))

print(resultado)
```

```python
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

resultado = list(filter(lambda p: p["precio"] > 100, productos))

print(resultado)
```
