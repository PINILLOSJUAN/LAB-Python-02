```python
def minimax(piedras, maximizando):

    if piedras == 0:
        if maximizando:
            return -1
        else:
            return 1

    if maximizando:
        mejor = -2

        for i in range(1, min(3, piedras) + 1):
            resultado = minimax(piedras - i, False)

            if resultado > mejor:
                mejor = resultado

        return mejor

    else:
        mejor = 2

        for i in range(1, min(3, piedras) + 1):
            resultado = minimax(piedras - i, True)

            if resultado < mejor:
                mejor = resultado

        return mejor


piedras = 10

print("JUEGO NIM")
print("Hay 10 piedras")
print("Puedes retirar 1, 2 o 3 piedras")
print("El humano comienza la partida")

while piedras > 0:

    print("\nPiedras restantes:", piedras)

    while True:
        humano = int(input("Cuantas piedras quieres retirar? "))

        if humano >= 1 and humano <= 3 and humano <= piedras:
            break
        else:
            print("Movimiento no valido, intenta nuevamente")

    piedras = piedras - humano

    print("El humano retiro", humano)

    if piedras == 0:
        print("Ganaste, la computadora no puede jugar")
        break

    print("\nTurno de la computadora")

    mejor = -2
    movimiento = 1

    for i in range(1, min(3, piedras) + 1):
        resultado = minimax(piedras - i, False)

        if resultado > mejor:
            mejor = resultado
            movimiento = i

    piedras = piedras - movimiento

    print("La computadora retiro:", movimiento)

    if piedras == 0:
        print("La computadora gano")
        break

print("\nFin del juego")
```
